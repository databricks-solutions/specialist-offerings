```python
import dlt

spark.conf.set('spark.databricks.cloudFiles.recordEventChanges', 'true')
spark.conf.set('spark.databricks.cloudFiles.optimizedEventSerialization.enabled','true')

@dlt.table(name='bronze')
def bronze():
  return (
    spark.readStream.format('cloudFiles')
    .option('cloudFiles.format', 'json')
    .option('maxFilesPerTrigger', 2)
    .load('/Users/alex.nastetsky@databricks.com/trips/original')
    .selectExpr("*", "_metadata as source_metadata")     
  )

@dlt.table(name='silver')
def silver():
  return (
    spark.readStream
    .table('LIVE.bronze')
  )

@dlt.table(name='gold')
def silver():
  return (
    spark.readStream
    .table('LIVE.silver')
  )
```
